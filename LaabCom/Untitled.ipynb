```python
risultati=[]

for x in range(1, 101):
    if x % 3 == 0 and x % 5 == 0: 
        risultati.append("MickeyMouse")
    elif x % 3 == 0: 
        risultati.append("Mickey")
    elif x % 5 == 0: 
        risultati.append("Mouse")
    else: risultati.append(x)

tupla_risultati=tuple(risultati)

nuovi=[]
for x in tupla_risultati:
    if x == "MickeyMouse":
        nuovi.append("DonaldDuck")
    elif x == "Mickey":
        nuovi.append("Donald")
    elif x == "Mouse":
        nuovi.append("Duck")
    else: nuovi.append(x) 
        
tupla_nuovi= tuple(nuovi)
     
```

```python
def swap(x, y):
    x, y = y, x
    return x, y

swap(2, 3)
```

```python
def distance (u,v):
    x1, y1= u
    x2, y2= v
    return ((x2-x1)**2+(y2-y1)**2)**(1/2)

x= 3, 0
y= 0, 4
distance(x, y)
    
```
